#### 58 Companies are being listed more than once.
#### Instead of deleting them lets create a combined list by doing arithmettic aggregation of multiple values.

```python
cs=df.groupby('Name').agg({
    'Ratings':'mean',
    'Count_Of_Rating':'mean',
    'Company_Types':'first',
    'Hq':'first',
    'Salary':'mean',
    'Interview_Counts':'mean',
    'Jobs':'mean',
    'Benefits':'mean'
}).reset_index()
cs.sort_values(by='Salary',ascending=False)
```

```python
rating=pd.cut(
    cs['Ratings'],
    bins=[0,1,2,3,4,5],
    labels=['0-1','1-2','2-3','3-4','4-5'],right=False
)
list=['0-1','1-2','2-3','3-4','4-5']
rating_count=rating.value_counts().reindex(list)
rating_count
```

```python
# Salary Distribution
salary_count,bins=pd.qcut(
    cs['Salary'],
    5,
    labels=['Very Low','Low','Medium','High','Very High'],
    retbins=True
)
temp=pd.DataFrame({
    'Category':salary_count.cat.categories,
    'Range': [f'{bins[i]:.2f}-{bins[i+1]:.2f}' for i in range (len(bins)-1)],
    'Count': salary_count.value_counts().sort_index()
    
    
})


#temp.drop(columns={'Percentage'},axis=1)
y=temp['Count'].sum()

temp['Percentage']=[(j/y)*100 for j in temp['Count']]
temp
```

```python
plt.figure(figsize=(15,6))
plt.pie(temp['Count'],labels=temp['Range']+ temp['Category'],autopct='%1.1f%%')
plt.title('Distribution of salary based on almost equal distribution of Company in each bucket')
```

```python
# Salary Distribution
salary_count,bins=pd.cut(
    cs['Salary'],
    bins=5,
    labels=['Very Low','Low','Medium','High','Very High'],
    retbins=True
)
temp1=pd.DataFrame({
    'Category':salary_count.cat.categories,
    'Range': [f'{bins[i]:.2f}-{bins[i+1]:.2f}' for i in range (len(bins)-1)],
    'Count': salary_count.value_counts().sort_index(),
    'Description':['Bottom 20 Percentile ','20-40 Percentile','40-60 Percentile','60-80 Percentile','Top 20 Percentile ']

    
})
y=temp1['Count'].sum()

temp1['Percentage']=[(j/y)*100 for j in temp1['Count']]
temp1
```

```python
plt.figure(figsize=(15,6))
plt.pie(temp1['Count'],labels=temp1['Category'],autopct='%1.1f%%')
plt.title('Distribution of cpmapnies based on equal segmentation of salary gap')
```
